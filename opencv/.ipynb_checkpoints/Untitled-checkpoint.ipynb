```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from IPython.display import Image

print(cv2.__version__)
```

```python
#load the image
image= cv2.imread("image\images.jpg" )
if image is not None:
    print("Image read sucessfully")
else:
    print("Failed to read the image")
    
```

```python
#display the image
img= cv2.resize(image,(600,400))
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#rewrite the image
img_new = img+30
cv2.imwrite("output.jpg" , img_new)
```
