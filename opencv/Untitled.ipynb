```python
#image fliping and crop
#fliping use for mirror the web cam view
import cv2
import matplotlib.pyplot as plt
img4 = cv2.imread("image/deerimage.jpg")
image_rgb = cv2.cvtColor(img4,cv2.COLOR_BGR2RGB)
plt.imshow(image_rgb)
print(img4)

# 1 -- >horizontal
# 0 -- >vertical
# -1 -- >both axes

flip_horizontal = cv2.flip(img4, 1)
flip_vertical = cv2.flip(img4, 0)
flip_both = cv2.flip(img4,-1)
titles = ["original", "horizontal", "vertical","both"]
images = [image_rgb,
cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB),
cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB),
cv2.cvtColor(flip_both,cv2.COLOR_BGR2RGB),
         ]

plt.figure(figsize=(12,8))

for i in range(4):
  plt.subplot(2,2,i+1)
  plt.imshow(images[i])
  plt.axis("off")

plt.tight_layout()
plt.show()




```
